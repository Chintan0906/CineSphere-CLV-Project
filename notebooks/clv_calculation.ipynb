```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

### Loading Master Feature Matrix

```python
MATRIX_FILE = 'C:/CLV-Project/data/cleaned_data_/master_feature_matrix.csv'
df_master = pd.read_csv(MATRIX_FILE)
```

### Computing Raw CLV Score

```python
df_master['clv_score'] = (
    df_master['total_ticket_spend'].fillna(0.0) +
    df_master['total_snack_spend'].fillna(0.0) +
    (df_master['avg_rating'].fillna(3.5) * 5.0)
).round(2)
```

### Normalizing CLV Score To 0-100 Scale

```python
scaler = MinMaxScaler(feature_range=(0, 100))
df_master['clv_normalized'] = scaler.fit_transform(df_master[['clv_score']]).round(2)
# Sort matrix by highest CLV score
df_master = df_master.sort_values('clv_score', ascending=False).reset_index(drop=True)
```

```python
clv_stats = df_master['clv_score'].describe()
print("\n" + "=" * 55)
print("📊 CLV TARGET VARIABLE DISTRIBUTION STATS")
print("=" * 55)
print(f"• Total Customers Audited : {int(clv_stats['count'])}")
print(f"• Mean CLV               : ${clv_stats['mean']:.2f}")
print(f"• Median CLV (50th %)    : ${clv_stats['50%']:.2f}")
print(f"• Minimum CLV            : ${clv_stats['min']:.2f}")
print(f"• Maximum CLV            : ${clv_stats['max']:.2f}")
print(f"• Standard Deviation     : ${clv_stats['std']:.2f}")
print("\n🏆 TOP 5 HIGHEST CLV CUSTOMERS:")
print(df_master[['username', 'total_ticket_spend', 'total_snack_spend', 'clv_score', 'clv_normalized']].head(5).to_string(index=False))
print("\n🔻 BOTTOM 5 LOWEST CLV CUSTOMERS:")
print(df_master[['username', 'total_ticket_spend', 'total_snack_spend', 'clv_score', 'clv_normalized']].tail(5).to_string(index=False))
```

```python
df_master.to_csv(MATRIX_FILE, index=False)
print("\n" + "=" * 55)
print(f"💾 Updated Master Matrix saved to {MATRIX_FILE}")
```
